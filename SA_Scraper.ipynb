```python
# Function to handle retries with exponential backoff
def fetch_data_with_retry(url, retries=3, delay=1):
    for attempt in range(retries):
        try:
            response = requests.get(url, timeout=10)  # Timeout in seconds
            if response.status_code == 200:
                return response.json()  # Return the JSON data if successful
            elif response.status_code == 429:
                retry_after = response.headers.get('Retry-After', 30)
                print(f"Rate limit exceeded. Retrying after {retry_after} seconds...")
                time.sleep(int(retry_after))
            else:
                print(f"Failed to fetch {url}. Status code: {response.status_code}")
                break
        except (ConnectionError, Timeout) as e:
            print(f"Connection error occurred: {e}. Retrying in {delay} seconds...")
            time.sleep(delay)  # Wait before retrying
            delay *= 2  # Exponential backoff
    return None  # Return None if all retries fail

# Function to construct the URL and fetch data for each id, cn pair
def fetch_data_for_id(id, cn):
    url = f'https://turkiye.servicesadvisor.net/api/services/item/{id}-{cn}?lang=en'
    json_data = fetch_data_with_retry(url)  # Get data with retry logic
    if json_data:
        data = json_data.get("data")
        if data:
            return data
    return None
```

```python


# List to store the results
all_data = []

# Set up parallel processing with ThreadPoolExecutor
with ThreadPoolExecutor(max_workers=10) as executor:  # Adjust max_workers as needed
    futures = []
    for id, cn in zip(df['id'], df['cn']):
        futures.append(executor.submit(fetch_data_for_id, id, cn))  # Submit tasks

    # Collect the results as they complete
    for future in as_completed(futures):
        data = future.result()
        if data:
            all_data.append(data)

# Convert the collected data into a DataFrame
sa_df = pd.DataFrame(all_data)

# Display the final DataFrame
sa_df
```
